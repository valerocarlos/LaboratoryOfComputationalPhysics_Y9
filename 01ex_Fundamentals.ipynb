## 2\. Convert the following function into a pure function with no global variables or side effects

````python
x = 5
def f1(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f1(alist)
print (ans)
print (alist) # alist has been changed!
````

## 3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def hello(func):
    def wrapper(*args, **kwargs):
        print("Hello! something is happening")
        return func(*args, **kwargs)
    return wrapper

@hello
def square(x):
    return x*x
for i in range(1,10):
    print(square(i))
````

## 4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
## a) WITH RECURSION

def factorial1(n):
   if n < 1:   # base case
       return 1
   else:
       return n * factorial1(n - 1)  # recursive call

factorial1(6)
````

````python
## b) WITHOUT RECURSION

def factorial2(n):
    base = 1 # case for 1 and 0. 
    for i in range(1, n + 1):
        base *= i #picks the current number and multiply it by the previous one and updates its value
    return base

factorial2(5)
````

````python
# check indeed the timing:

%timeit factorial1(20)
%timeit factorial2(20)
````

## 5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
import math

densities = {"Al": [0.5, 1, 2], "Fe": [3, 4, 5], "Pb": [15, 20, 30]}
radii = [1, 2, 3]
g = 9.8 # gravity  --> weight will be W = rho*f(r)*g where f(r) will be perimeter, area and volume of a circumference, circle and sphere

# Circumference, area and volume (simply the formulas for them)
functions = [lambda r: f(r) for f in [lambda r: 2*math.pi*r, lambda r: math.pi*r**2, lambda r: (4/3)*math.pi*r**3]]

for material, dens in densities.items(): #picks the material from densities list
    print(material)

    for radius in radii:
        weights = list(map(
            lambda pair: pair[0] * pair[1](radius)*g,
            zip(dens, functions) # this is the function that computes the weight by picking the first value of dens and the first value
            # of functions and gravity and multiply them, for each radius and element dnesity. 
        ))
        print(f"r = {radius}: {weights}")
````

````python
# Parent class
class Dog:

    # Class attribute
    species = 'mammal'

    # Initializer / Instance attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.is_hungry = True # initialize the instace that the dogs are hungry

    # Instance method
    def description(self):
        return "{} is {} years old".format(self.name, self.age)

    # Instance method
    def speak(self, sound):
        return "{} says {}".format(self.name, sound)

    # New method
    def eat(self):
        self.is_hungry = False


# Child classes
class RussellTerrier(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)


class Bulldog(Dog):
    def run(self, speed):
        return "{} runs {}".format(self.name, speed)
````

````python
# Create the dogs
my_dogs = [
    RussellTerrier("Lola", 6),
    Bulldog("Lia", 7),
    Dog("Frodo", 9)
]

print("I have {} dogs.".format(len(my_dogs)))

for dog in my_dogs:
    print("{} is {}.".format(dog.name, dog.age))

print("And they're all mammals, of course.")

# Feed every dog
for dog in my_dogs:
    dog.eat()

# Check whether all dogs have the same hunger status
if all(dog.is_hungry for dog in my_dogs):
    print("My dogs are hungry.")
elif all(not dog.is_hungry for dog in my_dogs):
    print("My dogs are not hungry.")
````
