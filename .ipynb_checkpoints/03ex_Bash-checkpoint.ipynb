### 1\.b Make two new files, one containing the students belonging to PoD, the other to Physics.

### 1\.c For each letter of the alphabet, count the number of students whose surname starts with that letter.

### 1\.d Find out which is the letter with most counts.

```python
%%bash

cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/students
max_count=0
max_letter=""
input="students_list.csv"

for letter in {A..Z}; do
    count=$(tail -n +2 "$input" | cut -d',' -f1 | grep -ic "^$letter")

    if (( count > max_count )); then
        max_count=$count
        max_letter=$letter
    fi
done

echo "Letter with the most students: $max_letter"
echo "Number of students: $max_count"
```

### 1\.e Assume an obvious numbering of the students in the file (first line is 1, second line is 2, etc.), group students "modulo 18", i.e. 1,19,37,.. 2,20,38,.. etc. and put each group in a separate file  

```python
%%bash

cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/students
input="students_list.csv"

# Create 18 files and copy the header into each one
for i in {1..18}; do
    head -n 1 "$input" > "group_${i}.csv"
done

# Number the students, skipping the header
tail -n +2 "$input" | while IFS= read -r line; do
    # Student numbering starts at 1
    student=$((student + 1))

    # Assign the student to a group from 1 to 18
    group=$(( (student - 1) % 18 + 1 ))

    echo "$line" >> "group_${group}.csv"
done
```

### FINAL SCRIPT FOR EX 1

```python
%%bash
set -e

# 1a: Download the CSV
cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9
mkdir -p Exercise1Bash
cd Exercise1Bash

input="students_list.csv"

wget -nv -O "$input" \
"https://www.dropbox.com/scl/fi/bxv17nrbrl83vw6qrkiu9/LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7&dl=1"

# Check that the downloaded file is not empty
if [[ ! -s "$input" ]]; then
    echo "Error: CSV download failed or the file is empty."
    exit 1
fi

echo "=== Download completed ==="
ls -lh "$input"


# 1b: Separate Physics and PoD students
head -n 1 "$input" > physics_students.csv
head -n 1 "$input" > pod_students.csv

tail -n +2 "$input" | while IFS= read -r line
do
    if [[ "$line" == *"Physics"* ]]; then
        echo "$line" >> physics_students.csv
    else
        echo "$line" >> pod_students.csv
    fi
done

echo
echo "=== Students by program ==="
echo "Physics students: $(( $(wc -l < physics_students.csv) - 1 ))"
echo "PoD students: $(( $(wc -l < pod_students.csv) - 1 ))"


# 1c: Count surnames starting with each letter
echo
echo "=== Students by surname initial ==="

for letter in {A..Z}; do
    count=$(tail -n +2 "$input" | cut -d',' -f1 | grep -ic "^${letter}" || true)
    echo "$letter: $count"
done


# 1d: Find the letter with the highest count
max_count=0
max_letter=""

for letter in {A..Z}; do
    count=$(tail -n +2 "$input" | cut -d',' -f1 | grep -ic "^${letter}" || true)

    if (( count > max_count )); then
        max_count=$count
        max_letter=$letter
    fi
done

echo
echo "=== Most common surname initial ==="
echo "Letter: $max_letter"
echo "Number of students: $max_count"


# 1e: Group students modulo 18
for i in {1..18}; do
    head -n 1 "$input" > "group_${i}.csv"
done

student=0

while IFS= read -r line; do
    student=$((student + 1))
    group=$(( (student - 1) % 18 + 1 ))

    echo "$line" >> "group_${group}.csv"
done < <(tail -n +2 "$input")

echo
echo "=== Modulo 18 groups ==="
for i in {1..18}; do
    count=$(( $(wc -l < "group_${i}.csv") - 1 ))
    echo "Group $i: $count students"
done

echo
echo "=== All tasks completed ==="
ls -lh
```

### Ex 2

### 2.a Make a copy of the file `data.csv` removing the metadata and the commas between numbers; call it `data.txt`

```python
%%bash

cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/Exercise2Bash
cp data.csv data.txt
grep -v "^#" data.txt > data_nometa.txt #create a temporary file with only the data without the metadata using grep
mv data_nometa.txt data.txt #overwrite the original txt
sed -i "s/,//g" data.txt #substitutes all the commas of the line (the g) with nothing in the input file
```

### 2\.b How many even numbers are there?

```python
%%bash
cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/Exercise2Bash
grep -oE '\b[0-9]*[02468]\b' data.txt > even_numbers.txt  #-oE for separating values on a row and extending pattern search
#it searches for elements in the line that end in 02468, so even numbers!
wc -l < even_numbers.txt
count=$(wc -l < even_numbers.txt)
echo "===Total number of even numbers===" >> even_numbers.txt
echo "$count" >> even_numbers.txt
```

### 2\.c Distinguish the entries on the basis of `sqrt(X^2 + Y^2 + Z^2)` is greater or smaller than `100*sqrt(3)/2`. Count the entries of each of the two groups 

```python
%%bash

cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/Exercise2Bash
awk '{
    norm = sqrt($1^2 + $2^2 + $3^2)

    if (norm >= 100*sqrt(3)/2)
        print $0, "--> Norm X,Y,Z greater than 100*sqrt(3)/2"
    else
        print $0, "--> Norm X,Y,Z smaller than 100*sqrt(3)/2"
}' data.txt > temp.txt
mv temp.txt data_classified.txt
```

### 2\.d Make `n` copies of data.txt (with `n` an input parameter of the script), where the i-th copy has all the numbers divided by i (with `1<=i<=n`).

```python
%%bash

cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/Exercise2Bash
n=10

for i in $(seq 1 "$n"); do
    awk -v divisor="$i" '{
        for (j=1; j<=NF; j++) #NF = Number of Fields, in this case number of columns on the file
            $j = $j / divisor
        print
    }' data.txt > "datacopy_${i}.txt"
done
```

## FINAL SCRIPT FOR EX 2

```python
%%bash

cd /home/charli/Desktop/LAB-COMP-PHYSICS/LaboratoryOfComputationalPhysics_Y9/Exercise2Bash || exit 1

# 1. Prepare data.txt: remove metadata and commas
cp data.csv data.txt
grep -v "^#" data.txt > data_nometa.txt
mv data_nometa.txt data.txt
sed -i "s/,//g" data.txt

# 2. Extract even numbers and count them
grep -oE '\b[0-9]*[02468]\b' data.txt > even_numbers.txt

count=$(wc -l < even_numbers.txt)
echo "Total number of even numbers: $count" >> even_numbers.txt

# 3. Classify data according to the norm of X, Y, Z
awk '{
    norm = sqrt($1^2 + $2^2 + $3^2)

    if (norm >= 100*sqrt(3)/2)
        print $0, "--> Norm X,Y,Z greater than or equal to 100*sqrt(3)/2"
    else
        print $0, "--> Norm X,Y,Z smaller than 100*sqrt(3)/2"
}' data.txt > data_classified.txt

# 4. Create n copies, dividing all numbers by i
n=10

for i in $(seq 1 "$n"); do
    awk -v divisor="$i" '{
        for (j=1; j<=NF; j++)
            $j = $j / divisor
        print
    }' data.txt > "datacopy_${i}.txt"
done

echo "All operations completed successfully!"
```
